# 1. What is Prompt Engineering?

**Simple definition (1 line):**
Prompt engineering is the skill of crafting clear, efficient inputs (prompts) that guide a large language model (LLM) to produce correct, useful, and cost-efficient outputs.

**Why the name?** It's two parts: *prompt* (the input you give the model) + *engineering* (designing that input deliberately, testing it, and optimizing it for quality and cost).

**Core idea:** LLMs are pre-trained on massive data. You cannot (usually) retrain them for every small task. Instead, you design prompts that coax the model to apply its existing knowledge in the exact way you want.

**Analogy:** Prompt engineering is like giving a precise blueprint to a skilled craftsman. The clearer the blueprint, the closer the final product will match your intent.

**Important notes:**
- Prompt engineering saves money (shorter, precise prompts reduce token usage) and reduces hallucination (by grounding and structure).
- It is evolving: new techniques (CoT, RAG, REACT, DSP) change how we use models.

**Quick example:**
- Weak prompt: `"Tell me about solar cells."`
- Strong prompt: `"In 5 bullet points suitable for a college-level student, explain how silicon solar cells convert light to electricity. Use simple analogies and list two disadvantages."`

The second prompt is more likely to produce a concise, useful answer tailored to the user.


# 2. Why learn Prompt Engineering?

- **High leverage skill:** With a single well-crafted prompt you can automate tasks that previously required code, reducing development time.

# 3. LLM Model Settings (what they mean and how to use them)

When using playgrounds or APIs you'll often see parameters. These matter.

**Model:** choose the model that fits accuracy vs cost (e.g., high-capacity models for hard reasoning, smaller models for routine text).

**Temperature (0.0 - ~1.2):** randomness of output.
- 0.0–0.2: very deterministic (best for factual or exact outputs).
- 0.5–0.8: creative but coherent.
- >1.0: very creative, but may lose coherence.

**Top-p (nucleus sampling):** consider words from the smallest set whose cumulative probability ≥ p.
- Lower top-p → more focused output; higher → more diverse.

**Max tokens (max-length):** maximum output size. Controls verbosity & cost.

**Stop sequences:** a list of tokens/strings that tell the model to stop generating when encountered. Useful for structured outputs.

**Frequency penalty:** discourages repeating the same token often. Good for stylistic variation.

**Presence penalty:** discourages using tokens that already appeared anywhere in the output. Useful to avoid reintroducing the same topics.

**Output format (e.g., JSON):** forcing a format helps downstream parsing, but requires careful prompt engineering (example templates below).

**Practical combos:**
- Precise factual tasks: temp=0.0, top_p=0.3–0.7, low penalties.
- Creative writing: temp=0.8–1.0, top_p=0.9, low penalties.
- Structured JSON output: temp=0.0–0.2, include example JSON and stop sequences.

**Remember:** settings change behavior — always test multiple settings and pick the one that balances quality and cost.

# 5. Prompt Engineering Techniques — Zero/One/Few-Shot

**Zero-shot:** Ask the model to perform a task with no examples.
- Use when the task is simple or when you want the model to use general knowledge.
- Example: `"Translate the sentence to Spanish: 'Good morning'"`.

**One-shot:** Provide one example to show the expected format or style.
- Useful when one example clarifies the mapping.
- Example: `"Example: 'Hello' -> 'Hola'. Now translate 'Good evening'."`

**Few-shot:** Provide several examples to demonstrate a pattern.
- Best for nuanced mappings or when the model needs to learn a pattern quickly.
- Example: show 3–5 input-output pairs then ask for the next mapping.

**When to use which?**
- Use few-shot for complex or unusual transformations.
- Use one-shot for small clarifications.
- Use zero-shot for quick checks or common knowledge tasks.

**Trade-offs:**
- More examples → better accuracy but higher token cost.
- Example quality matters more than quantity (clear, diverse examples are best).

# 7. Self-Consistency in Prompt Engineering

**Definition:** Self-consistency means the model gives stable and repeatable answers for the same prompt.

**Why it matters:**
- Inconsistent answers reduce trust.
- In production, deterministic behavior may be required for predictable features.

**How to improve self-consistency:**
1. **Lower temperature & top-p** to reduce randomness.
2. **Provide examples** (few-shot) to anchor expected format.
3. **Use CoT with voting:** generate multiple CoT outputs, then take majority or vote for the most common final answer (this is a research technique to improve accuracy).
4. **Fine-tuning** (if available): train the model on domain-specific data to standardize outputs.

**Practical trick — majority voting with CoT:**
- Generate N different CoT outputs (with low-medium temp).
- Extract the final answers and pick the most frequent one.

**Note:** generating multiple outputs costs more tokens, but increases reliability on hard reasoning tasks.

# 9. Role Playing in Prompt Engineering

**What it is:** Tell the model to adopt a persona (teacher, interviewer, doctor, historian). It helps get style and perspective.

**Why use role-play:**
- Achieve consistent tone and behavior.
- Simulate interviews, dialogues, and domain experts.

**Prompt pattern:**
```
You are [ROLE] with [experience/constraints]. Answer as [tone]. Example: 'You are a senior backend engineer with 10 years experience. Explain microservices to a junior developer in 3 bullets.'
```

**Examples:**
- `"You are a patient doctor. Explain this diagnosis gently and list next steps."`
- `"You are an HR interviewer. Ask 6 behavioral questions for a front-end developer."`

**Tip:** Combine role-play + constraints + examples for best results.

# 11. REACT — Reasoning + Acting framework

**What REACT captures:** Combine **RE**asoning and **ACT**ion. The model first reasons (often with CoT), then performs actions (like writing code, calling APIs, or returning a structured result).

**Why useful:**
- Useful for multi-step interactive tasks where the model must *decide* and *act*.
- Encourages the model to separate analysis from final output.

**Example prompt pattern:**
```
Step 1 (Reason): Explain your plan step-by-step. Step 2 (Act): Execute the final step and produce the result enclosed in <RESULT> tags.
```

**Use-cases:**
- Interactive agents, chatbots that call external tools, step-by-step data transformation.

**Caveat:** REACT-style prompts increase token usage (explicit reasoning) but often improve reliability.

# Revision Tips & One-page Cheat Sheet (Read this one day before interview)

**Quick mental checklist for each prompt:**
1. Objective — what do I want? (1 sentence)
2. Role — set a persona if needed.
3. Format — JSON, bullets, length, example.
4. Examples — 0/1/few-shot? Add if needed.
5. LLM Settings — temp, top_p, max_tokens, stop sequences.
6. Safety/Constraints — banned topics, citations.

- Factual: `temp=0.0`, `top_p=0.3-0.7`.
- Creative: `temp=0.7-1.0`, `top_p=0.9`.
- Structured JSON: `temp=0.0-0.2`, use examples + stop sequence.

**One-line prompts you can memorize:**
- Summarize long text: `"Summarize in 5 bullets, each <=20 words."`
- Create interview questions: `"You are an HR interviewer. Provide 8 behavioral questions for a backend engineer."`
- Convert to JSON: `"Output only valid JSON with keys: ..."`

**Quick debugging steps:**
- If output is off-format: add a clear example.
- If model hallucinates: lower temp, include evidence, or use RAG.
- If repetitive: increase frequency/presence penalty.

**Final tip:** Practice writing prompts with small tests and keep a library of templates.
